```python
import pandas as pd

known_variable_average_list = ['stress_score', 'spo2', 'filteredDemographicVO2Max',
                               'lightly_active_minutes', 'moderately_active_minutes',
                               'very_active_minutes', 'sleep_duration', 'minutesAsleep']
known_variable_sum_list = ['distance', 'steps', 'calories']
all_vars = [c for c in known_variable_average_list + known_variable_sum_list
            if c in df_daily.columns]

key = ["id", "date"]

# 1. Overall: how many rows share the same user + date?
sizes = df_daily.groupby(key).size()
print("Total rows:", len(df_daily))
print("Unique (id, date) pairs:", len(sizes))
print("Pairs with more than 1 row:", (sizes > 1).sum())
print("Rows that are part of a duplicate pair:", df_daily.duplicated(subset=key, keep=False).sum())
print("\nHow many pairs have 1, 2, 3... rows:")
print(sizes.value_counts().sort_index())

# 2. Per variable: among duplicated (id, date) pairs, how many have
#    - more than one non-missing value
#    - more than one DISTINCT value (i.e., the duplicates disagree)
dup = df_daily[df_daily.duplicated(subset=key, keep=False)]
g = dup.groupby(key)

summary = pd.DataFrame({
    "pairs_with_multiple_values": {v: (g[v].count() > 1).sum() for v in all_vars},
    "pairs_with_conflicting_values": {v: (g[v].nunique() > 1).sum() for v in all_vars},
})
print("\nDuplicated pairs per variable:")
print(summary)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

x_vars = ['lightly_active_minutes', 'moderately_active_minutes', 'very_active_minutes',
          'sleep_duration', 'minutesAsleep', 'calories']
# y_good_vars = [ 'HAPPY', 'NEUTRAL', 'ENTERTAINMENT']
# y_bad_vars = ['ALERT','SAD', 'TENSE/ANXIOUS', 'TIRED']

y_good_vars = [ 'HAPPY_next', 'NEUTRAL_next', 'ENTERTAINMENT_next']
y_bad_vars = ['ALERT_next','SAD_next', 'TENSE/ANXIOUS_next', 'TIRED_next']

def plot_moods(data, x_vars, y_vars, title, palette_name="tab10"):
    palette = dict(zip(y_vars, sns.color_palette(palette_name, len(y_vars))))

    fig, axes = plt.subplots(2, 3, figsize=(16, 8), sharey=True)
    for ax, x in zip(axes.ravel(), x_vars):
        for y in y_vars:
            ax.scatter(data[x], data[y], s=15, alpha=0.5, color=palette[y], label=y)
        ax.set_xlabel(x)
        ax.grid(alpha=0.3)

    axes[0, 0].set_ylabel("Mood count")
    axes[1, 0].set_ylabel("Mood count")

    handles, labels = axes[0, 0].get_legend_handles_labels()
    fig.legend(handles, labels, loc="center right", title="Mood", markerscale=2)
    fig.suptitle(title)
    fig.tight_layout(rect=[0, 0, 0.9, 0.96])
    plt.show()

plot_moods(df_train, x_vars, y_good_vars, "Positive / neutral moods", palette_name="Set1")
plot_moods(df_train, x_vars, y_bad_vars,  "Negative / activated moods", palette_name="tab10")
```

```python
import numpy as np
from sklearn.model_selection import KFold
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.metrics import root_mean_squared_error

mood_train = df_train
num_splits = 5
num_models = 5          # was 4; added the quadratic model
kfold = KFold(num_splits, random_state=216, shuffle=True)

known_columns = ['lightly_active_minutes', 'moderately_active_minutes',
                 'very_active_minutes', 'sleep_duration', 'minutesAsleep', 'calories']

for column in next_cols:
    rmses = np.zeros((num_models, num_splits))
    i = 0
    predicted_columns = [column]
    print("evaluation", column)
    for train_index, test_index in kfold.split(mood_train):
        mood_tt = mood_train.iloc[train_index]
        mood_ho = mood_train.iloc[test_index]

        ## Baseline
        mean_regressor = DummyRegressor(strategy="mean")
        mean_regressor.fit(mood_tt[known_columns], mood_tt[predicted_columns])
        rmses[0, i] = root_mean_squared_error(
            mood_ho[predicted_columns], mean_regressor.predict(mood_ho[known_columns]))

        ## MLR
        mlr = LinearRegression()
        mlr.fit(mood_tt[known_columns], mood_tt[predicted_columns])
        rmses[2, i] = root_mean_squared_error(
            mood_ho[predicted_columns], mlr.predict(mood_ho[known_columns]))

        ## KNN
        knn = KNeighborsRegressor(10)
        knn.fit(mood_tt[known_columns], mood_tt[predicted_columns])
        rmses[3, i] = root_mean_squared_error(
            mood_ho[predicted_columns], knn.predict(mood_ho[known_columns]))

        ## Quadratic model (squares + pairwise interactions of all features)
        quad = make_pipeline(
            StandardScaler(),
            PolynomialFeatures(degree=2, include_bias=False),
            LinearRegression(),
        )
        quad.fit(mood_tt[known_columns], mood_tt[predicted_columns])
        rmses[4, i] = root_mean_squared_error(
            mood_ho[predicted_columns], quad.predict(mood_ho[known_columns]))

        i = i + 1

    print(f"Baseline Avg. CV RMSE: {np.mean(rmses[0,:])} and STD: {np.std(rmses[0,:])}")
    print(f"MLR Avg. CV RMSE: {np.mean(rmses[2,:])} and STD: {np.std(rmses[2,:])}")
    print(f"neighbour regressor CV RMSE: {np.mean(rmses[3,:])} and STD: {np.std(rmses[3,:])}")
    print(f"Quadratic Avg. CV RMSE: {np.mean(rmses[4,:])} and STD: {np.std(rmses[4,:])}", "\n")
```

```python
# do classification test:
if MERGE_DAILY==False:
    import numpy as np
    from sklearn.model_selection import StratifiedKFold
    from sklearn.dummy import DummyClassifier
    from sklearn.linear_model import LogisticRegression
    from sklearn.neighbors import KNeighborsClassifier
    from sklearn.pipeline import make_pipeline
    from sklearn.preprocessing import PolynomialFeatures, StandardScaler
    from sklearn.metrics import accuracy_score, f1_score, roc_auc_score

    mood_train = df_train
    num_splits = 5
    num_models = 4
    kfold = StratifiedKFold(num_splits, random_state=216, shuffle=True)

    known_columns = ['lightly_active_minutes', 'moderately_active_minutes',
                    'very_active_minutes', 'sleep_duration', 'minutesAsleep', 'calories']

    model_names = ["Baseline (majority)", "Logistic", "KNN", "Quadratic logistic"]

    for column in next_cols:          # each mood column is 0/1
        accs = np.zeros((num_models, num_splits))
        f1s  = np.zeros((num_models, num_splits))
        aucs = np.zeros((num_models, num_splits))
        print("evaluation", column)

        # StratifiedKFold needs y so each fold keeps the 0/1 ratio
        y = mood_train[column]
        i = 0
        for train_index, test_index in kfold.split(mood_train, y):
            mood_tt = mood_train.iloc[train_index]
            mood_ho = mood_train.iloc[test_index]

            X_tt, y_tt = mood_tt[known_columns], mood_tt[column]   # note: 1D y, not [column]
            X_ho, y_ho = mood_ho[known_columns], mood_ho[column]

            models = [
                DummyClassifier(strategy="prior"),
                make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)),
                make_pipeline(StandardScaler(), KNeighborsClassifier(10)),
                make_pipeline(StandardScaler(),
                            PolynomialFeatures(degree=2, include_bias=False),
                            LogisticRegression(max_iter=5000)),
            ]

            for m, model in enumerate(models):
                model.fit(X_tt, y_tt)
                pred = model.predict(X_ho)
                prob = model.predict_proba(X_ho)[:, 1]
                accs[m, i] = accuracy_score(y_ho, pred)
                f1s[m, i]  = f1_score(y_ho, pred, zero_division=0)
                aucs[m, i] = roc_auc_score(y_ho, prob)
            i += 1

        for m, name in enumerate(model_names):
            print(f"{name:20s} Acc {accs[m].mean():.3f}±{accs[m].std():.3f} | "
                f"F1 {f1s[m].mean():.3f}±{f1s[m].std():.3f} | "
                f"AUC {aucs[m].mean():.3f}±{aucs[m].std():.3f}")
        print()
```

```python
# do classification test:
# predict new users
if MERGE_DAILY == False:
    import numpy as np
    from sklearn.model_selection import StratifiedGroupKFold
    from sklearn.dummy import DummyClassifier
    from sklearn.linear_model import LogisticRegression
    from sklearn.neighbors import KNeighborsClassifier
    from sklearn.pipeline import make_pipeline
    from sklearn.preprocessing import PolynomialFeatures, StandardScaler
    from sklearn.metrics import accuracy_score, f1_score, roc_auc_score

    mood_train = df_train
    num_splits = 5
    num_models = 4
    kfold = StratifiedGroupKFold(n_splits=num_splits, shuffle=True, random_state=216)

    known_columns = ['lightly_active_minutes', 'moderately_active_minutes',
                     'very_active_minutes', 'sleep_duration', 'minutesAsleep', 'calories']

    model_names = ["Baseline (majority)", "Logistic", "KNN", "Quadratic logistic"]

    groups = mood_train["id"]        # each user stays entirely in train or in holdout

    for column in next_cols:         # each mood column is 0/1
        accs = np.full((num_models, num_splits), np.nan)
        f1s  = np.full((num_models, num_splits), np.nan)
        aucs = np.full((num_models, num_splits), np.nan)
        print("evaluation", column)

        y = mood_train[column]
        i = 0
        for train_index, test_index in kfold.split(mood_train, y, groups=groups):
            mood_tt = mood_train.iloc[train_index]
            mood_ho = mood_train.iloc[test_index]

            X_tt, y_tt = mood_tt[known_columns], mood_tt[column]
            X_ho, y_ho = mood_ho[known_columns], mood_ho[column]

            # a fold with a single class in train or holdout can't be fit or scored
            if y_tt.nunique() < 2 or y_ho.nunique() < 2:
                print(f"  fold {i}: skipped (single class)")
                i += 1
                continue

            # sanity check: no user appears in both train and holdout
            assert set(mood_tt["id"]).isdisjoint(mood_ho["id"])

            models = [
                DummyClassifier(strategy="prior"),
                make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)),
                make_pipeline(StandardScaler(), KNeighborsClassifier(10)),
                make_pipeline(StandardScaler(),
                              PolynomialFeatures(degree=2, include_bias=False),
                              LogisticRegression(max_iter=5000)),
            ]

            for m, model in enumerate(models):
                model.fit(X_tt, y_tt)
                pred = model.predict(X_ho)
                prob = model.predict_proba(X_ho)[:, 1]
                accs[m, i] = accuracy_score(y_ho, pred)
                f1s[m, i]  = f1_score(y_ho, pred, zero_division=0)
                aucs[m, i] = roc_auc_score(y_ho, prob)
            i += 1

        for m, name in enumerate(model_names):
            print(f"{name:20s} Acc {np.nanmean(accs[m]):.3f}±{np.nanstd(accs[m]):.3f} | "
                  f"F1 {np.nanmean(f1s[m]):.3f}±{np.nanstd(f1s[m]):.3f} | "
                  f"AUC {np.nanmean(aucs[m]):.3f}±{np.nanstd(aucs[m]):.3f}")
        print()
```
